```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv(r"C:\Users\vedan\Downloads\archive (1)\Student_Performance.csv")
```

```python
df.head()
```

```python
df['Extracurricular Activities']=df['Extracurricular Activities'].map({"Yes":1, "No":0})
```

```python
df.head()
```

```python
x=df[['Hours Studied','Previous Scores','Extracurricular Activities','Sleep Hours','Sample Question Papers Practiced']]
y=df['Performance Index']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.75,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lrmodel=LinearRegression()
lrmodel.fit(x_train,y_train)
```

```python
x_pred=lrmodel.predict(x_train)
y_pred=lrmodel.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
r2=np.sqrt(mse)
mae
mse
r2
```

```python
inter=lrmodel.intercept_
coef=lrmodel.coef_
inter
```

```python
coef
```

```python
print(f"Performance Index = {inter:.2f} + ({coef[0]:.2f} * Hours Studied) + ({coef[1]:.2f}* Previous Scores) + ({coef[2]:.2f} * Extracurricular Activities) + ({coef[3]:.2f} * Sleep Hours) + ({coef[4]:.2f} * Sample Papers Practices)")
```

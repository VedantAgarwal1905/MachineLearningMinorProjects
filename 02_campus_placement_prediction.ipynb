```python
data['Innovative Project(Y/N)']=data['Innovative Project(Y/N)'].replace('yes','Yes')
```

```python
data['Technical Course(Y/N)'].value_counts()
```

```python
data['Technical Course(Y/N)']=data['Technical Course(Y/N)'].replace('Yess','Yes')
```

```python
data.head()
```

```python
#now i will be removing email and name columns from the dataframe because they are irrelevant for training thee model
data.drop(['Email','Name'],axis=1,inplace=True) 
```

```python
# now i will be perfroming encodding on the categorical features... i will perfrom binary encoding on features that have only 2 unique values and i will be performing one hot encodig on the features that have more than 2 values

binarycols=['Internships(Y/N)','Training(Y/N)','Innovative Project(Y/N)','Technical Course(Y/N)','Backlog in 5th sem']
for col in binarycols:
    data[col]=data[col].map({'Yes':1, 'No':0})
    
data['Gender']=data['Gender'].map({'Male':1,'Female':0})

data['Placement(Y/N)?']=data['Placement(Y/N)?'].map({'Placed':1,'Not Placed':0})
```

```python
data=pd.get_dummies(data,columns=['10th board','12th board','Stream'],dtype=int)
```

```python
data.shape
```

```python
x=data.drop(['Placement(Y/N)?'],axis=1)
y=data['Placement(Y/N)?']
```

```python
x.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.75,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
x_train=scale.fit_transform(x_train)
x_test=scale.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
logi_model=LogisticRegression()
logi_model.fit(x_train,y_train)
```

```python
y_pred=logi_model.predict(x_test)
```

```python
from sklearn.metrics import classification_report, accuracy_score
```

```python
logi_cr=classification_report(y_test,y_pred)
accuracy=accuracy_score(y_test,y_pred)
accuracy
```

```python
print(logi_cr)
```

```python
#now we are gonna prepare the model for svm 
```

```python
from sklearn.svm import LinearSVC
svm_model=LinearSVC()
svm_model.fit(x_train,y_train)
```

```python
y_pred_svm=svm_model.predict(x_test)
```

```python
svm_accuracy=accuracy_score(y_test,y_pred_svm)
```

```python
svm_accuracy
```

```python
svm_classification=classification_report(y_test,y_pred_svm)
```

```python
print(svm_classification)
```

```python
#now we are gonna perform the sgd classification
```

```python
from sklearn.linear_model import SGDClassifier
```

```python
sgd_model=SGDClassifier()
sgd_model.fit(x_train,y_train)
```

```python
y_pred_sgd=sgd_model.predict(x_test)
```

```python
sgd_classification=classification_report(y_test,y_pred_sgd)
print(sgd_classification)
```

```python
sgd_accuracy=accuracy_score(y_test,y_pred_sgd)
print(sgd_accuracy)
```
